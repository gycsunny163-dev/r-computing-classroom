## Definitions and conditions
mean() 为算术平均值；sd() 为样本标准差，分母是 n-1。标准差描述观测值离散程度；标准误在适当独立同分布假设下描述均值估计的抽样不确定性，两者不能混用。小样本或系统偏差不能仅靠 sd 消除。
