## Definitions and conditions
function 定义函数；参数接收输入。return() 显式返回值；没有 return() 时返回最后一个表达式的值。计算函数不要只打印结果，应返回可继续用于运算的值。

## Worked example · exposed
以下是教材示例，不是你的独立作答。

```r
distance_at <- function(speed_m_s, time_s) {
  return(speed_m_s * time_s)
}
print(distance_at(2, 3))
```
