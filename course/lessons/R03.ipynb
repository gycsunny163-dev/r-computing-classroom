## Definitions and conditions
typeof() 检查底层类型；普通 2 通常是 double，2L 是 integer。字符与数字不同。NA 表示缺失，不是 0；is.na() 检查缺失。排除缺失会改变样本，必须说明；转换失败可能产生 warning 和 NA。

## Worked example · exposed
以下是教材示例，不是你的独立作答。

```r
x <- c(2, NA, 6)
print(typeof(x))
print(is.na(x))
print(mean(x))
print(mean(x, na.rm = TRUE))
```
