## Definitions and conditions
lm(y ~ x, data=...) 拟合含截距的线性模型；coef() 给出系数，residuals() 是观测值减拟合值。普通最小二乘最小化残差平方和。斜率单位为 y单位/x单位。拟合或高相关不证明因果，也不消除测量偏差。

## Worked example · exposed
以下是教材示例，不是你的独立作答。

```r
measurements <- read.csv("../course/data/linear_motion.csv")
fit <- lm(distance_m ~ time_s, data=measurements)
print(coef(fit))
print(residuals(fit))
```
